```python
# ========================================
# 1️⃣ Importy i konfiguracja
# ========================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import glob
import os

# Styl wykresów
sns.set(style="whitegrid", palette="muted", font_scale=1.1)

# ========================================
# 2️⃣ Wczytanie i połączenie wszystkich CSV
# ========================================
data_path = "../data/raw/"
file_paths = glob.glob(os.path.join(data_path, "*.csv"))

print(f"Znaleziono {len(file_paths)} plików CSV:")
for f in file_paths:
    print(" -", os.path.basename(f))

dfs = [pd.read_csv(f) for f in file_paths]
df = pd.concat(dfs, ignore_index=True)

print(f"\nŁączny rozmiar danych: {df.shape}")
display(df.head())

# ========================================
# 3️⃣ Podstawowe informacje o danych
# ========================================
print("\n🔍 Informacje o typach danych:")
display(df.info())

print("\n📊 Statystyki opisowe:")
display(df.describe(include="all").transpose())

# ========================================
# 4️⃣ Braki danych
# ========================================
missing = df.isnull().sum().sort_values(ascending=False)
missing = missing[missing > 0]

if not missing.empty:
    print("\n⚠️ Braki danych:")
    display(missing)
    plt.figure(figsize=(10,5))
    sns.barplot(x=missing.index, y=missing.values)
    plt.xticks(rotation=45)
    plt.title("Liczba brakujących wartości na kolumnę")
    plt.show()
else:
    print("\n✅ Brak brakujących danych!")

# ========================================
# 5️⃣ Zmienne numeryczne
# ========================================
num_cols = df.select_dtypes(include=[np.number]).columns
print(f"\n🔢 Zmienne numeryczne ({len(num_cols)}): {list(num_cols)}")

for col in num_cols:
    plt.figure(figsize=(10,4))
    plt.subplot(1,2,1)
    sns.histplot(df[col], kde=True)
    plt.title(f"Rozkład: {col}")
    plt.subplot(1,2,2)
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot: {col}")
    plt.tight_layout()
    plt.show()

# ========================================
# 6️⃣ Zmienne kategoryczne
# ========================================
cat_cols = df.select_dtypes(exclude=[np.number]).columns
print(f"\n🔠 Zmienne kategoryczne ({len(cat_cols)}): {list(cat_cols)}")

for col in cat_cols:
    plt.figure(figsize=(8,4))
    df[col].value_counts().head(15).plot(kind='bar')
    plt.title(f"Top wartości w kolumnie: {col}")
    plt.ylabel("Liczność")
    plt.xticks(rotation=45)
    plt.show()

# ========================================
# 7️⃣ Korelacje między zmiennymi numerycznymi
# ========================================
if len(num_cols) > 1:
    plt.figure(figsize=(10,8))
    corr = df[num_cols].corr()
    sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", square=True)
    plt.title("Mapa korelacji")
    plt.show()

# ========================================
# 8️⃣ Pairplot dla wybranych kolumn
# ========================================
if len(num_cols) > 3:
    sample_cols = num_cols[:4]
    print(f"\n📈 Pairplot dla kolumn: {sample_cols}")
    sns.pairplot(df[sample_cols], diag_kind="kde")
    plt.show()

# ========================================
# 9️⃣ Duplikaty i wartości odstające
# ========================================
duplicates = df.duplicated().sum()
print(f"\n📋 Liczba duplikatów: {duplicates}")

# prosta detekcja odstających
from scipy import stats
z_scores = np.abs(stats.zscore(df.select_dtypes(include=[np.number]), nan_policy='omit'))
outliers = (z_scores > 3).any(axis=1).sum()
print(f"⚠️ Liczba potencjalnych wartości odstających (Z-score > 3): {outliers}")

print("\n✅ EDA zakończone.")
```
